```python
# Import necessary libraries
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
```

```python
# ---------------------------
# Step 1: Load the Iris Dataset
# ---------------------------
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['target'] = iris.target
df['species'] = df['target'].map({0: 'setosa', 1: 'versicolor', 2: 'virginica'})  # Map target to species name
```

```python
# ---------------------------
# Step 2: Explore the Data Structure
# ---------------------------
print("\n📌 First 5 rows of the dataset:")
print(df.head())

print("\n📌 Dataset Information:")
print(df.info())

print("\n📌 Summary Statistics:")
print(df.describe())
```

```python
# ---------------------------
# Step 3: Check for Missing Values
# ---------------------------
print("\n📌 Checking for missing values:")
print(df.isnull().sum())  # No missing values in Iris dataset
```

```python
# ---------------------------
# Step 4: Visualize Feature Distributions
# ---------------------------
# Histogram for each feature
df.drop(columns=['target']).hist(figsize=(10, 6), bins=15, color='skyblue', edgecolor='black')
plt.suptitle("Feature Distributions (Histogram)", fontsize=14)
plt.show()

# Boxplot for outlier detection
plt.figure(figsize=(10, 6))
sns.boxplot(data=df.drop(columns=['target']), palette="Set2")
plt.title("Boxplot of Features for Outlier Detection")
plt.xticks(rotation=45)
plt.show()
```

```python
# ---------------------------
# Step 5: Pair Plot to Explore Relationships
# ---------------------------
sns.pairplot(df, hue="species", palette="husl", diag_kind="kde")
plt.suptitle("Pairplot of Iris Features", y=1.02)
plt.show()
```

```python
# ---------------------------
# Step 6: Correlation Heatmap
# ---------------------------
plt.figure(figsize=(8,6))
sns.heatmap(df.drop(columns=['target', 'species']).corr(), annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Feature Correlation Heatmap")
plt.show()
```
